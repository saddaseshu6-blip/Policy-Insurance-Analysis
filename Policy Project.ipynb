```python
import zipfile
import pandas as pd
```

```python
zip_path = 'Policy_Data_All_Sheets_CSV.zip'

with zipfile.ZipFile(zip_path, 'r') as z:
    data = {}

    for file in z.namelist():
        if file.endswith('.csv'):
            sheet_name = file.split('/')[-1].replace('.csv', '')
            data[sheet_name] = pd.read_csv(z.open(file))
```

```python
df= pd.merge(
    data['Policy_Details'],
    data['Customer_Information'],
    on = 'Customer ID',
    how = 'left'
)
```

```python
claims_summary = data['Claims'].groupby('Policy ID').agg(
    Total_Claims=('Claim ID', 'count'),
    Total_Claim_Amount=('Claim Amount', 'sum')
).reset_index()

df = pd.merge(
    df,
    claims_summary,
    on='Policy ID',
    how='left'
)
```

```python
df.shape
```

```python
Payment_Summary = data['Payment_History'].groupby('Policy ID').agg(
    Total_Payments = ('Payment ID', 'count'),
    Total_Amount_Paid = ('Amount Paid','sum')
).reset_index()
df=pd.merge(
    df,
    Payment_Summary,
    on = 'Policy ID',
    how ='left'
)
```

```python
data['Customer_Information']['Age Group']  = np.select(
    [
        data['Customer_Information']['Age'] <=27,
        data['Customer_Information']['Age'] <=37,
        data['Customer_Information']['Age'] <=47,
        data['Customer_Information']['Age'] <=57,
        data['Customer_Information']['Age'] <=67,
        data['Customer_Information']['Age'] <=77,
        data['Customer_Information']['Age'] <=87,
        data['Customer_Information']['Age'] <=99
    ],
    [
        '18-27',
        '28-37',
        '38-47',
        '48-57',
        '58-67',
        '68-77',
        '78-87',
        '88-99'
    ],
    default = 'Other'
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize = (10,6))

ax= sns.violinplot (
    x='Status',
    y='Risk Score',
    data=df
)

medians =df.groupby('Status')['Risk Score'].median()

for i,Status in enumerate(medians.index):ax.text(
    i,
    medians[Status]+3,
    f'median:{medians[Status]:.1f}',
    ha='center',
    fontweight='bold',
     color='black'
    )
plt.title('Risk Score Distribution by Status')
plt.xlabel('Status')
plt.ylabel('Risk Score')

plt.tight_layout()
plt.show()
```

```python
df['Policy End Date'] = pd.to_datetime(
    df['Policy End Date'],
    errors='coerce'
)

time_trend = df.groupby('Policy End Date')['Policy ID'].count()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

time_trend.plot(kind='line', marker='o')

plt.title('Policy End Date - Time Series Trend')
plt.xlabel('Policy End Date')
plt.ylabel('Number of Policies')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
monthly_trend = (
    df.set_index('Policy End Date')
      .resample('ME')['Policy ID']
      .count()
)

monthly_trend.plot(figsize=(12,5), marker='o')

plt.title('Monthly Policy End Date Trend')
plt.xlabel('Month')
plt.ylabel('Number of Policies')
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
df['Policy End Date'] = pd.to_datetime(
    df['Policy End Date'],
    errors='coerce'
)

Quarter_trend = (
    df.set_index('Policy End Date')
    .resample('Q')['Policy ID'].count()
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize =(12,5))

quarterly_trend.plot(kind='line',marker ='0')
plt.title('Quarterly Policy End Date Trend')
plt.xlabel('Quarter')
plt.ylabel('Number of Policies')
plt.grid(True)
plt.xticks(rotation =45)
plt.tight_layout()
plt.show()
```
